```python
# Report Card Generator using Dictionary

print("===== REPORT CARD GENERATOR =====")

# Student details
name = input("Enter student name: ")
roll_no = input("Enter roll number: ")

# Dictionary to store subject marks
marks = {}

subjects = ["Math", "Science", "English", "History", "Computer"]

# Input marks
for subject in subjects:
    score = float(input(f"Enter marks for {subject}: "))
    marks[subject] = score

# Calculate total
total = sum(marks.values())

# Calculate percentage
percentage = total / len(subjects )




# Grade calculation
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

# Display report card
print("\n===== REPORT CARD =====")
print(f"Name       : {name}")
print(f"Roll No    : {roll_no}")
print("----------------------------")

for subject, score in marks.items():
    print(f"{subject:10}: {score}")

print("----------------------------")
print(f"Total Marks: {total}")
print(f"Percentage : {percentage:.2f}%")
print(f"Grade      : {grade}")

# Pass/Fail
result = "Pass" if grade != "F" else "Fail"
print(f"Result     : {result}")

print("============================")
```
